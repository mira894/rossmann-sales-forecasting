```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")

df = pd.read_csv("../data/processed/rossmann_base.csv", parse_dates=["Date"])
df.shape
```

```python
plt.figure(figsize=(10,4))
df.groupby("Date")["Sales"].sum().plot()
plt.title("Total Daily Sales Over Time")
plt.ylabel("Sales")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="DayOfWeek", y="Sales", data=df)
plt.title("Sales Distribution by Day of Week")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="Promo", y="Sales", data=df)
plt.title("Impact of Promotions on Sales")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(x="StoreType", y="Sales", data=df)
plt.title("Sales by Store Type")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.select_dtypes(include=["int64", "float64"]).corr(), cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

### Key EDA Insights

- Sales show clear weekly seasonality.
- Promotions significantly increase median sales but increase variance.
- Certain store types consistently outperform others.
- Sales distribution is right-skewed, justifying robust metrics like MAE.
